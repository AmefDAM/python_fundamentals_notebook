## 1. Variables and Assignment

A variable is a name that refers to an object in Python.

You create a variable by assigning a value to a name using the `=` operator.

```python
age = 25
name = "Alice"
height = 1.68
is_student = True

print(age)
print(name)
print(height)
print(is_student)
```

In this example:

- `age` refers to an integer.
- `name` refers to a string.
- `height` refers to a floating-point number.
- `is_student` refers to a Boolean value.

Python determines the type of the object from the assigned value.

```python
print(type(age))
print(type(name))
print(type(height))
print(type(is_student))
```

## 2. Reassigning Values

A variable can be reassigned to refer to a different value.

The new assignment replaces the previous binding of the name.

```python
score = 10
print(score)

score = 20
print(score)
```

After the second assignment, `score` refers to the integer `20`.

You can also use the current value of a variable to calculate its next value.

```python
score = 10

score = score + 5
print(score)

score = score * 2
print(score)
```

In `score = score + 5`, Python first evaluates the expression on the right-hand side.

It then assigns the result to `score`.

This pattern is frequently used to update counters and accumulators in algorithms.

## 3. Initializing Counters and Accumulators

A counter tracks how many times something happens.

An accumulator collects or combines values as an algorithm runs.

Both usually need an initial value before they can be updated.

Common initial values include:

| Purpose | Initial value |
|---|---|
| Sum or total | `0` |
| Count | `0` |
| Product | `1` |
| String construction | `""` |
| List construction | `[]` |

The initial value depends on the operation being performed.

### 3.1. Counters

A counter is commonly increased by one whenever a condition is met.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_count = 0

for number in numbers:
    if number % 2 == 0:
        even_count += 1

print(even_count)
```

The variable `even_count` starts at `0`.

Each time an even number is found, the counter increases by one.

The final value represents the number of even elements in the list.

### 3.2. Accumulators

An accumulator combines values as an algorithm processes them.

For example, a running total can be calculated by adding each number to an accumulator.

```python
numbers = [1, 2, 3, 4, 5]

total = 0

for number in numbers:
    total += number

print(total)
```

The accumulator starts at `0`, which is the appropriate initial value for a sum.

Each iteration adds the current number to `total`.

When the loop finishes, `total` contains the sum of all the numbers.

## 4. Augmented Assignment Operators

Augmented assignment operators combine an operation with an assignment.

For example:

`total += number`

is equivalent to:

`total = total + number`

These operators are useful when updating counters, accumulators and other variables.

```python
value = 10

value += 5
print(value)

value -= 3
print(value)

value *= 2
print(value)

value /= 4
print(value)
```

Other augmented assignment operators include:

| Operator | Example | Meaning |
|---|---|---|
| `+=` | `x += 2` | Add and assign |
| `-=` | `x -= 2` | Subtract and assign |
| `*=` | `x *= 2` | Multiply and assign |
| `/=` | `x /= 2` | Divide and assign |
| `//=` | `x //= 2` | Floor-divide and assign |
| `%=` | `x %= 2` | Calculate the remainder and assign |
| `**=` | `x **= 2` | Raise to a power and assign |

Note that augmented assignment is not always identical to writing the longer expression, particularly for mutable objects.

```python
number = 17

number //= 5
print(number)

number = 17
number %= 5
print(number)

number = 3
number **= 2
print(number)
```

## 5. Descriptive Names and Common Mistakes

### 5.1. Choose Descriptive Names

Variable names should communicate the purpose of the stored value.

Descriptive names make code easier to understand, debug and maintain.

Python conventionally uses `snake_case` for variable names.

```python
# Less descriptive
x = [10, 20, 30]
s = 0

for n in x:
    s += n

print(s)
```

```python
# More descriptive
numbers = [10, 20, 30]
total = 0

for number in numbers:
    total += number

print(total)
```

Both examples produce the same result, but the second makes the purpose of each variable clearer.

Prefer names such as `total`, `even_count`, `price` and `student_name` when they accurately describe the data.

### 5.2. Assignment vs. Equality

The assignment operator (`=`) assigns a value to a name.

The equality operator (`==`) compares two values and returns a Boolean result.

These operators have different purposes.

```python
number = 10

print(number == 10)
print(number == 5)
```

A common mistake is confusing `=` with `==`.

Use `=` to assign a value and `==` to compare values, for example in an `if` condition.

```python
number = 10

if number == 10:
    print("The number is 10")
```

### 5.3. Initialize Variables Before Updating Them

A variable must have a value available in the current scope before it can be used in an expression.

For example, trying to increase an undefined variable raises a `NameError`.

```python
# This raises a NameError if total has not been defined:
not_defined_total += 5
```

Initialize the accumulator before the loop.

```python
not_defined_total = 0 #now defined
not_defined_total += 5

print(not_defined_total)
```

### 5.4. Avoid Shadowing Built-in Names

Avoid using names such as `sum`, `list`, `str` or `max` for your own variables.

These names are also used by Python's built-in functions and types.

Reusing them can make the original built-in inaccessible through that name in the current scope.

```python
numbers = [1, 2, 3]

total = 0

for number in numbers:
    total += number

print(total)
print(sum(numbers))
```

Use names such as `total`, `numbers` and `result` instead of replacing built-in names unnecessarily.

## 6. Practical Application: Summing Values in a List

A common programming task is calculating the sum of all the values in a list.

The general algorithm is:

1. Initialize an accumulator to `0`.
2. Iterate over the list.
3. Add each number to the accumulator.
4. Return or display the final total.

This pattern is useful in many algorithmic problems.

```python
numbers = [3, 7, 2, 8]

total = 0

for number in numbers:
    total += number

print(total)
```

### 6.2. Sum Only Positive Values

Sometimes an algorithm must accumulate only values that satisfy a condition.

For this example, add a number to the total only when it is greater than zero.

Adding zero would not change the sum, so checking `number > 0` is sufficient for this task.

```python
numbers = [-5, 3, -2, 8, 0, 4]

total = 0

for number in numbers:
    if number > 0:
        total += number

print(total)
```

### 6.3. Turn the Algorithm into a Function

Once the algorithm works, it can be encapsulated in a function so it can be reused with different lists.

This is the basic idea behind the `positive_sum` Codewars kata.

```python
def positive_sum(numbers):
    total = 0

    for number in numbers:
        if number > 0:
            total += number

    return total


print(positive_sum([-5, 3, -2, 8, 0, 4]))
print(positive_sum([1, 2, 3, 4, 5]))
print(positive_sum([-1, -2, -3]))
```

The function combines several concepts from this notebook:

- Variable initialization: `total = 0`
- Iteration: `for number in numbers`
- Conditional execution: `if number > 0`
- Augmented assignment: `total += number`
- Returning a result: `return total`

The variable `total` is an accumulator because it collects the sum of the values that satisfy the condition.
