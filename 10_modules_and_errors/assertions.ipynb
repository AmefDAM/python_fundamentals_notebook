# Assertions en Python

## ¿Qué es una assertion?

Una *assertion* es una comprobación que verifica que una condición sea verdadera.

En Python se utiliza la sentencia `assert`:

```python
assert 2 + 3 == 5
```

Si la expresión es verdadera, la ejecución continúa sin producir ningún resultado visible.

Si es falsa, Python lanza una excepción `AssertionError`:

```python
assert 2 + 3 == 6
```

Esto provoca un error porque `2 + 3 == 6` es falso.

## Sintaxis

```python
assert condicion
```

También se puede incluir un mensaje explicativo:

```python
assert edad >= 0, "La edad no puede ser negativa"
```

Si la condición es falsa, el mensaje se incluye en la excepción `AssertionError`.

## Ejemplos

### Comprobar igualdad

```python
result = 2 + 3

assert result == 5
```

### Comprobar desigualdad

```python
assert result != 0
```

### Comprobar una condición booleana

```python
is_valid = True

assert is_valid
```

### Comprobar pertenencia

```python
languages = ["Python", "Java", "SQL"]

assert "Python" in languages
```

### Comprobar que un valor no sea None

```python
result = "completed"

assert result is not None
```

## assert en pruebas automatizadas

Frameworks como pytest utilizan las sentencias `assert` para comprobar los resultados esperados.

```python
def add(a, b):
    return a + b


def test_add():
    assert add(2, 3) == 5
```

En este ejemplo, la prueba pasa si `add(2, 3)` devuelve `5`.

La sentencia `assert` expresa la condición que debe cumplirse. pytest se encarga de descubrir y ejecutar la función de prueba, además de informar del resultado.

Consulta la documentación de [pytest](../../development-environment-notes/testing/pytest.md) para aprender a organizar y ejecutar pruebas.

## Diferencia entre assert y raise

Aunque ambos pueden provocar una excepción, tienen finalidades distintas.

### assert

Se utiliza para comprobar una condición que se espera que sea verdadera, especialmente durante el desarrollo y en pruebas.

```python
assert total >= 0
```

### raise

Se utiliza para lanzar explícitamente una excepción cuando el programa debe comunicar o gestionar una situación de error.

```python
if total < 0:
    raise ValueError("El total no puede ser negativo")
```

No conviene utilizar `assert` para validar datos externos o condiciones que deban comprobarse siempre en producción: Python puede desactivar las assertions cuando se ejecuta con optimizaciones (`-O`).

Para validar entradas y comunicar errores de ejecución, utiliza comprobaciones normales y excepciones apropiadas.

## Ideas clave

- `assert` comprueba que una expresión sea verdadera.
- Si la condición es falsa, se lanza `AssertionError`.
- Puede añadirse un mensaje después de una coma.
- pytest utiliza `assert` para comprobar resultados en las pruebas.
- No debe sustituir a la validación de datos ni al manejo de errores de producción.
